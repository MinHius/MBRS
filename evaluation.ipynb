```python
import matplotlib.pyplot as plt

# Mapping attacks to their parameter labels and whether to invert the x-axis
# (Inverted so x-axis consistently reads: Least Aggressive -> Most Aggressive)
attack_config = {
    "Crop": {
        "x_col": "severity",
        "x_label": "Retained Area (%)",
        "scale_fn": lambda x: x * x * 100,
        "invert_x": True,  # 1.0 (no crop) -> 0.1 (extreme crop)
    },
    "Cropout": {
        "x_col": "severity",
        "x_label": "Retained Area (%)",
        "scale_fn": lambda x: x * x * 100,
        "invert_x": False,  # 0.1 (tiny box) -> 1.0 (full wipe)
    },
    "Dropout": {
        "x_col": "severity",
        "x_label": "Retained Area (%)",
        "scale_fn": lambda x: x * 100,
        "invert_x": True,  # 1.0 (keep 100%) -> 0.1 (keep 10%)
    },
}

for attack_name, group in traditional_attacks_df.groupby("attack"):
    cfg = attack_config.get(
        attack_name,
        {
            "x_col": "severity",
            "x_label": "Resized Area (%)",
            "scale_fn": lambda x: x * 100,
            "invert_x": True,
        },
    )

    df_plot = group.sort_values(cfg["x_col"])

    x_vals = cfg["scale_fn"](df_plot[cfg["x_col"]])
    y_vals = df_plot["BER"] * 100

    plt.figure(figsize=(8, 5))
    plt.plot(
        x_vals,
        y_vals,
        marker="o",
        color="green",
        label=f"{attack_name} BER",
    )

    if cfg["invert_x"]:
        plt.gca().invert_xaxis()

    plt.xlabel(cfg["x_label"])
    plt.ylabel("BER (%)")
    plt.title(f"DWT-DCT - {attack_name} Robustness")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```
